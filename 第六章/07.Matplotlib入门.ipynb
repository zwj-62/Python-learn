```python
import random

import matplotlib.pyplot as plt
plt.rcParams['font.sans-serif'] = ['SimHei']#设置可以显示中文
# #目标：绘制折线统计图
# x=[2, 5, 8, 12, 15, 19, 22, 25, 27, 30]
# y=[2, 5, 8, 12, 15, 19, 22, 25, 27, 30]
# plt.plot(x,y)#折线图
# plt.show()#显示图
# x=[2, 5, 8, 12, 15, 19, 22, 25, 27, 30]
# y=[random.randint(2,30) for i in x]
# plt.plot(x,y)
# plt.show()




x=[i for i in range(1,31)]
y_jx=[random.randint(25,34) for i in x]
y_yn=[random.randint(15,30) for i in x]
plt.figure(figsize=(10,5))
plt.plot(x,y_jx,label='嘉兴')#----->如果没有画布则创建一个画布，否则自动创建
plt.plot(x,y_yn,label='云南')
#设置折线图的详细信息
plt.title('浙江嘉兴气温变化的折线图',fontsize=15)
plt.xlabel('时间')#x轴刻度
plt.ylabel('温度')#y轴刻度
plt.xticks(x[::])#x轴刻度
y_ticks=[i for i in range(15,41)]
plt.yticks(y_ticks)#y轴刻度
plt.grid(linestyle='--')
plt.legend(loc='upper right')
plt.show()
```
