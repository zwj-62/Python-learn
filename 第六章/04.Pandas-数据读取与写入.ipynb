```python
import pandas as pd
#读取数据  --read_csv
df=pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'])
df
#数据的处理
df['销售金额']=df['销售数量']*df['单价']
df
#写入文件 --to_csv
df.to_csv('data/sales_01.csv',index=False)#index= Flase 表示索引列不写入我们的数据



#数据查看--->head tail describe info shape colums
df1=pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'])
#head 显示sales中前10行的数据，如果括号里不写数据，那么就默认读前五行
df1.head(10)
#tail 显示sales中后10行的数据，如果括号里不写数据，那么就默认读后五行
df1.tail(10)
#describe 是对sales中可以统计的信息进行统计,也就输对数据进行处理，显示数据的统计信息
df1.describe()
#info 展示的是sales的数据信息等（列明，非空计数，数据类型等）
df1.info
#shape 展示数据的行数和列数
df1.shape
#colums 显示数据的列名，以及类型
df1.columns




#数据选择 基本上是选择列或选择行
#1.选择行
df2=pd.read_csv('data/sales.csv',usecols=['订单号','产品类别','产品名称','销售数量','单价'],index_col='订单号')
#可以直接使用df2['']的形式显示某一列的信息 也可以使用df2.列的名称 比如bf2.产品名称
df2['产品名称']

#如果要显示多列信息，则使用的方法是df2[['','','']]
df2[['产品名称','单价']]
#2.选择列 --iloc,loc
#iloc 基于行号选择行，语法df.iloc[start:stop:step]包好start，不包含stop
df2.iloc[0:5:1]
df2.iloc[0:6:2]
df2
#loc--->基于行索引选择（包括结束位置），语法为df.loc[start:stop:step]
df2.loc[0:10:1]#这种就不行，因为我们现在索引是订单号，所以要从订单号到订单号
df2.loc[5800000000:5800001238:1]
df2.loc[5800000000:5800001238:2]

#数据的过滤
df3=pd.read_csv('data/sales.csv')
df3
#数据过滤------>df[过滤的条件]
#1.获取 销售>=10的订单数据
df3[df3['销售数量']>=10]
#2.获取产品类别为食品或图书的订单数据 用isin可以筛选类别中的内容
df3[df3['产品类别'].isin(['食品','图书'])]
#3.获取单价在100-200之间的订单数据 范围用between 默认包含边界
df3[df3['单价'].between(100,200)]
#4.获取销售数量>=8,并且单价>=100的订单数据------>多条件并且用& 或者用|
df3[(df3['销售数量']>=8) & (df3['单价']>=100)] #多个条件需要用括号隔开
#5.获取产品类别为服装/食品，支付方式为支付宝/微信支付的订单数量
df3[(df3['产品类别'].isin(['服装','食品']))&df3['支付方式'].isin(['支付宝','微信支付'])]

```
