```python
import pandas as pd
#读取数据
df=pd.read_csv('data/sales.csv',nrows=10)
#排序
#根据 销售数量 倒序排序
df.sort_values('销售数量',ascending=False)
#根据 价格 升序排序
df.sort_values('单价',ascending=True)
#根据价格 升序排序 价格一样的情况再根据 销售数量 倒叙排序
df.sort_values(['单价','销售数量'],ascending=[True,False])


df=pd.read_csv('data/sales.csv',nrows=50)
df
#数据分组
#根据 产品类别分组 ，统计各个类别的订单数量---->count()
df.groupby('产品类别')['订单号'].count()
##根据产品类别分组，统计各个类别的销售数量之和----》sum
df.groupby('产品类别')['销售数量'].sum()
#根据产品类别分组，统计各个类别的销售金额之和
df['销售金额']=df['单价']*df['销售数量']#先加入销售金额这一列
df.groupby('产品类别')['销售金额'].sum()
##根据产品类别分组，统计各个类别的最高商品 单价 -max
df.groupby('产品类别')['单价'].min()
##根据产品类别分组，统计各个类别的最高商品 单价 -max
df.groupby('产品类别')['单价'].max()
##根据产品类别分组，统计各个类别的平均商品 单价 -mean
df.groupby('产品类别')['单价'].mean()
##根据产品类别分组，统计各个类别的平均商品单价 最高单价 最低单价
df.groupby('产品类别')['单价'].aggregate(['mean','max','min'])
###根据产品类别分组，统计各个类别的销售数量之和，销售金额之和，平均单价
df.groupby('产品类别').agg({'销售数量':'sum','销售金额':'sum','单价':'mean'})
```
