```python
import matplotlib.pyplot as plt
from matplotlib.axes import Axes
from matplotlib.pyplot import figure

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']#中文字体设置为黑体
#创建子图
#figure :画布对象，axes:子图数组
figure ,axes=plt.subplots(nrows=1, ncols=2,figsize=(20,6),dpi=100)  #nrows 行 ncols 列
#柱状图 图一
coutries=['中国','德国','美国','英国','俄罗斯','哈萨克斯坦','伊朗']
values=[45,33,20,34,10,24,51]
axes1:Axes =axes[0]
axes1.set_title('世界石油储备',fontsize=20)#h绘制标题
axes1.bar(coutries,values,color='green',width=0.8)#绘制柱状图  width:柱子宽度，color 柱子颜色
axes1.set_xlabel('国家',fontsize=18)
axes1.set_ylabel('石油储备：亿吨',fontsize=18)
axes1.grid(linestyle='-.',alpha=0.5)


#图二 饼状图
coutries = ['中国', '美国', '日本', '德国', '法国', '巴西', '澳大利亚', '印度', '埃及', '加拿大']
values = [14.1, 3.4, 1.24, 0.84, 0.68, 2.16, 0.27, 14.4, 1.13, 0.40]
axes2:Axes=axes[1]
axes2.pie(values,labels=coutries,autopct='%1.1f%%')
axes2.set_title('世界人口的国家占比',fontsize=18)
axes2.legend(loc='lower center', ncol=5,bbox_to_anchor=(0.5, -0.1),fontsize=12  )#显示图例，ncol是显示图例的行数，bbox是显示图例的位置（0.5是控制水平方位，-0.1是控制垂直方位
# ）
#保存图片

plt.savefig('data/01.png')
```
