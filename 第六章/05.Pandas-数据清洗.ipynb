```python
import pandas as pd
pd.set_option('display.min_rows', 50)
#读取数据
df=pd.read_csv('data/sales.csv')
#2数据清洗

# #2.1缺失值的处理----》先查看是否有缺失值，
# df.isnull()
# #2.1.1删除缺失值所在行 列
# df.dropna()
# df.dropna(axis=1)
# df
# #2.1.2填充缺失值
df.fillna('安徽')
# df.ffill()#填充缺失值，用上一行数据填充缺失值
# df.bfill()#填充缺失值，用下一行的数据填充缺失值
#2.2重复值处理
#2.2.1查看重复值
df.duplicated()#查看所有列都重复的重复值
df.duplicated(subset=['订单号'])#查看订单号重复的订单

#2.2.2删除重复值
df.drop_duplicates(subset=['订单号'], keep="first")#keep ='last'保留重复的最后一行，‘first’保留重复的第一行，Flase 都不保留

#2.3异常值处理
#2.3.1查看异常值
df[df['单价']<0]
#2.3.2 可以直接删除异常值所在的行
df.drop(df[df['单价']<0].index)
#2.3.3也可以修复异常值，比如说改成绝对值
df['单价']=df['单价'].abs()#让单价这列数据全部变成绝对值
df
#2.4数据格式转换
df['订单日期']=df['订单日期'].str.replace('/',"-")
df
```
