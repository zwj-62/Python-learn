```python
## 练习一：8 行订单，验证每一步是否生效

import pandas as pd

df = pd.DataFrame([
    [1001, '食品', '饼干', 2, 10, '杭州', '2026-10-01'],
    [1002, '食品', '牛奶', 3, 20, None, '2026/10/02'],
    [1002, '食品', '牛奶', 3, 20, '北京', '2026-10-02'],
    [1003, '电子产品', '鼠标', 1, -50, '宁波', '2026-10-03'],
    [1004, '图书', 'Python入门', -2, 40, '杭州', '2026-10-04'],
    [1005, '图书', 'Python入门', 2, 40, None, '2026-10-05'],
    [1006, '电子产品', '键盘', 1, 100, '宁波', '2026-10-06'],
    [1007, '食品', '咖啡', 0, 30, '上海', '2026-10-07'],
], columns=[
    '订单号', '产品类别', '产品名称', '销售数量',
    '单价', '客户所在城市', '订单日期',
])

# 需求按顺序完成：
#
# 1. 输出每列缺失数量、订单号的重复数量、单价和数量非正的记录数量。
df.isnull().sum()
df.duplicated(subset='订单号').sum()
(df['单价']<0).sum()
(df['销售数量']<=0).sum()
print("每列缺失的数量：\n",df.isnull().sum())
print("订单号的重复数量：",(df.duplicated(subset='订单号').sum()))
print("单价非正数量：",(df['单价']<=0).sum())
print("销售数量非正数量：",(df['销售数量']<=0).sum())
# 2. 仅将“客户所在城市”的缺失值填为“未知城市”。
df['客户所在城市']=df['客户所在城市'].fillna("未知城市")
# 3. 本练习约定一行一单。按订单号去重，保留输入中的第一条记录。
df.duplicated(subset='订单号')
df=df.drop_duplicates(subset='订单号', keep='first')
# 4. 本练习仅统计正常销售。排除单价小于等于 0 或销售数量小于等于 0 的记录，不取绝对值。
df=df.drop(df[df['单价']<=0].index)
df=df.drop(df[df['销售数量']<=0].index)
df
# 5. 统一日期分隔符。日期类型转换是补充项：可用 `pd.to_datetime(..., format='%Y-%m-%d')` 继续转为日期类型。
df['订单日期']=df['订单日期'].str.replace('/','-')
df
# 6. 计算“销售金额”，等于单价乘以销售数量。
df['销售金额']=df['单价']*df['销售数量']
df
# 7. 按产品类别统计订单数、总销售数量、总销售金额和平均单价，并按总销售金额降序排列。
# df.groupby('产品类别').agg({'订单号':'count','销售数量':'sum','销售金额':'sum',' 单价':'mean'})
# df.sort_values('销售金额',ascending=False)
# #按产品类别进行分组，分组之后计算各类的订单数、总销售数量、总销售金额和平均单价
catagory_report=df.groupby('产品类别').agg({
    '订单号':'count',
    '销售数量':'sum',
    '销售金额':'sum',
    '单价':'mean'
})
#对分类汇总结果按照总销售金额降序排列
catagory_report=catagory_report.sort_values('销售金额',ascending=False)
catagory_report
```
