```python
# ## 练习二：正常销售订单清洗与分类报表
# 用途：把今天的方法连成一份可以放到 GitHub 上展示的数据处理练习。项目质量以能独立解释和重复运行为准，不需要增加界面。
# 1. 读取整个文件，不使用 `nrows=50` 限制数据量。
import pandas as pd
df = pd.read_csv(r'D:\APP\python\python-test\Python\第六章\data\sales.csv')
# 2. 客户城市缺失时填“未知城市”。
df['客户所在城市']=df['客户所在城市'].fillna('未知城市')
df
# 3. 本练习按一行一单处理，按订单号去重，保留原始文件中的第一条记录。
df.drop_duplicates(subset=['订单号'], keep='first', inplace=True)
# 4. 去重后，仅保留单价大于 0 且销售数量大于 0 的正常销售记录。
df=df[(df['销售数量']>0)&(df['单价']>0)]
df
# 5. 统一日期分隔符；选做：将日期列转成日期类型。
df['订单日期']=df['订单日期'].str.replace('/', '-')
df
# 6. 在清洗后的数据上计算销售金额。
df['销售金额']=df['销售数量']*df['单价']
df
# 7. 按产品类别统计订单数、总销量、总销售金额、平均单价；按总销售金额从大到小排列。
df_category=df.groupby('产品类别').agg({
    '订单号': 'count',
    '销售数量': 'sum',
    '销售金额': 'sum',
    '单价': 'mean'
})
df_category.sort_values(by='销售金额', ascending=False, inplace=True)
df_category
#把清洗后的数据放入新文件sales_02.csv
df.to_csv(r'D:\APP\python\python-test\Python\第六章\data\sales_02.csv', index=False)
# 8. 导出分类报表到category_report.csv。
df_category.to_csv(r'D:\APP\python\python-test\Python\第六章\data\category_report.csv', index=True)
```
